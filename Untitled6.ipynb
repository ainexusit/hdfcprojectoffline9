```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['A', 'B', 'C'],
    'Math': [85, 90, 88],
    'Science': [92, 85, 91]
})

print(df)
```

```python
melted_df = pd.melt(
    df,
    id_vars=['Name'],
    var_name='Subject',
    value_name='Marks'
)
print(melted_df)
```

```python
import pandas as pd

df = pd.DataFrame({
    'dept': ['IT','IT','IT','HR','HR'],
    'salary': [90000, 90000, 85000, 70000, 70000]
})
```

```python
df['row_num' ] =df['salary'].index+1
```

```python
df['rank'] = df['salary'].rank(method='min')
```

```python
df['dense_rank'] = df['salary'].rank(method='dense')
```

```python
df['r1']=df.groupby('dept')['salary'].rank(method='min')
```
